```python
import numpy as np

arr = np.random.rand(6, 3)
transposed = arr.T

print(arr.shape)         # (6, 3)
print(transposed.shape)  # (3, 6)
print(transposed)
```

```python
import numpy as np

n, p, m = 2, 3, 4

A = np.random.rand(n, p)   # n x p
B = np.random.rand(p, m)   # p x m

result = np.dot(A, B)      # or A @ B
print(result)
print(result.shape)        # (n, m) = (2, 4)
```
